## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
import pandas as pd

df = pd.read_csv('../../data/raw/content_refresh_anonymized.csv') 

active_df = df[df['impressions_90d'] > 100]

total_active_pages = active_df.shape[0]

ctr_by_type = active_df.groupby('content_type')['ctr'].mean().sort_values()

print(f"1. We have {total_active_pages} active pages with significant impressions to analyze.")
print(f"2. The lowest performing content type has an average CTR of: {ctr_by_type.min():.4f}")
print(f"3. The highest performing content type has an average CTR of: {ctr_by_type.max():.4f}")
```
