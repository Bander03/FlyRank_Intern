## 1. My lane as an ML task (type)

*Classification, clustering, ranking, or scoring — which one, and why?*

```python
My project falls under the Scoring / Regression task type. Instead of placing items into rigid categories (classification) or sorting a list (ranking), I am training a model to predict a continuous numeric performance score—specifically the Click-Through Rate (CTR)—based on the textual attributes and categorical classifications of a page.
```

## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*

```python
The target variable I want to predict is ctr (Click-Through Rate). If the raw ctr is too sparse or volatile for certain low-impression pages, I will use a smoothed version of CTR or a weighted engagement score as a proxy to ensure the model captures stable, meaningful performance trends.
```

## 3. Success metric

*One metric you can defend. What number means 'good'?*

```python
Technical Success Metric: I will use Mean Absolute Error (MAE) and R-squared ($R^2$) to evaluate how close the model's predicted performance scores are to the actual observed scores.
Business Success Metric: The model is successful if it can identify underperforming or overperforming text structures more accurately than a random baseline, allowing the content team to make data-backed writing decisions.
```

## 4. The unit of analysis, as a real dataframe

*Load your lane's slice and show it: one row = one what?*

```python
import pandas as pd

# Load the content dataset from raw files
df = pd.read_csv('../../data/raw/content_refresh_anonymized.csv')

# Clean and establish the unit of analysis (Active Pages)
active_pages_df = df[df['impressions_90d'] > 100].copy()

# Print the unit of analysis definition
print("--- Unit of Analysis ---")
print("One row in this dataframe represents ONE unique web page (URL) and its performance metrics over a 90-day window.")
print(f"Total active rows (pages) to analyze: {active_pages_df.shape[0]}\n")

# Display a preview of the unit of analysis with the target column
preview_cols = ['content_type', 'content_age_days', 'avg_position', 'impressions_90d', 'ctr']
active_pages_df[preview_cols].head()
```

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

```python
A fixed rule like "long-form content always gets a higher CTR" is too rigid because human search behavior is complex and non-linear. Text performance changes drastically depending on its rank (avg_position) and its structural style (content_type). A machine learning model beats a fixed rule because it can learn multi-dimensional, non-linear interactions between position, page age, and textual attributes simultaneously, adjusting its expectations dynamically rather than relying on a static assumption.
```
